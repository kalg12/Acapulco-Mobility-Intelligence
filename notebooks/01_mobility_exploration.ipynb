```python
import pandas as pd

data = {
    "route": [
        "Centro - Costera",
        "Centro - Costera",
        "Centro - Diamante",
        "Centro - Diamante",
        "Renacimiento - Centro",
        "Renacimiento - Centro",
    ],
    "transport": [
        "Taxi",
        "Platform",
        "Taxi",
        "Platform",
        "Taxi",
        "Platform",
    ],
    "price_mxn": [
        100, 85,
        250, 210,
        180, 160,
    ],
}

df = pd.DataFrame(data)

df
```

```python
df.groupby("transport")["price_mxn"].mean()
```

```python
comparison = df.pivot_table(
    index="route",
    columns="transport",
    values="price_mxn",
    aggfunc="mean"
)

comparison
```

```python
comparison.plot(
    kind="bar",
    figsize=(10, 5),
    title="Transportation Fare Comparison — Simulated Data"
)

plt.xlabel("Route")
plt.ylabel("Price (MXN)")
plt.xticks(rotation=30, ha="right")
plt.tight_layout()
plt.show()
```
